```python
from pathlib import Path
```

```python
LATEX_HEADER = """
\\documentclass{article}
\\usepackage[paperwidth=36cm, paperheight=20cm, margin=1mm]{geometry}
\\usepackage{graphicx}
\\usepackage{underscore}
\\usepackage[sfdefault]{roboto}
\\begin{document}
\\pagestyle{empty}
"""
```

```python
new_files = sorted(Path("../new_kernel_figures").glob("*.pdf"))
latex_filename = "plot_comparison.tex"
with open(latex_filename, "w", encoding="utf-8") as latex_file:
    latex_file.writelines(LATEX_HEADER)
    for new_file in new_files:
        old_file = Path("../figures") / new_file.name
        latex_file.write(f"{new_file.name}\\par\n")
        latex_file.write(
            "\\includegraphics{%s}\n\\includegraphics{%s}\n\\clearpage\n\n"
            % (new_file, old_file)
        )
    latex_file.writelines("\\end{document}\n")
```
