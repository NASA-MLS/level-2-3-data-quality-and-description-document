```python
import git
import subprocess
```

```python
# Open the log file and extract the relevant part
with open("v6-x-report.log", "r", encoding="utf-8", errors="ignore") as log_file:
    log_lines = [line.strip() for line in log_file.readlines()]
```

```python
i_start = log_lines.index("*File List*")
i_end = log_lines.index("***********")
used_files = [line.split(" ")[0] for line in log_lines[i_start + 1 : i_end]]
```

```python
local_files = []
for file in used_files:
    result = subprocess.run(["kpsewhich", file],
                           check=True, capture_output=True, text=True)
    full_path = result.stdout.strip()  # kpsewhich prints the path
    local_files.append(full_path)
```
