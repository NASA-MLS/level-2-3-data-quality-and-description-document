```python
# This notebook looks to see which files that git is managing aren't used by LaTeX
```

```python
import subprocess
from pathlib import Path
from git import Repo
```

```python
# Open the fls file and extract the relevant lines
with open("v6-x-report.fls", "r", encoding="utf-8", errors="ignore") as fls_file:
    fls_lines = [line.strip() for line in fls_file.readlines()]
prefix = "INPUT "
used_files = [file[len(prefix) :] for file in fls_lines if file.startswith(prefix)]
```

```python
# Now skip all the ones that come from texlive
for prefix in ["/usr/local/texlive/", "/Users/livesey/Library/texlive"]:
    used_files = [file for file in used_files if not file.startswith(prefix)]
```

```python
# Remove all the "./" prefixes
used_files = [file.replace("./", "") for file in used_files]
# remove duplicates and sort
used_files = set(used_files)
```

```python
# Now work out what files are managed by git
# point at the repo root (Path.cwd() if you run inside the repo)
repo = Repo(Path.cwd())
# All tracked, non-ignored files (relative paths, one per line)
tracked_files = set(repo.git.ls_files().splitlines())
```

```python
tracked_but_not_used = sorted(tracked_files - used_files)
```

```python
# Now work out which ones it's OK to have in this state
whitelist = [
    ".gitattributes",
    ".gitignore",
    "announcement.txt",
    "makediffs.sh",
    "mybib.bib",
    ".vscode/settings.json",
    "list-unchanged.sh",
    "notes-05feb15a.txt",
    "shortnames.bib",
    "tracking.txt",
    "v6-x-report.synctex(busy)",
    "valpaperstable.tex",
    "check_unused_files.ipynb",
]
tracked_and_not_wanted = []
for file in tracked_but_not_used:
    # Keep files that are whitelisted
    if file in whitelist:
        continue
    # Keep eps files if their pdf forebears are used
    if file.endswith(".eps") and file.replace(".eps", ".pdf") in used_files:
        continue
    # Keep the averaging kernel text files
    if file.startswith("figures/avk-") and file.endswith(".txt"):
        continue
    tracked_and_not_wanted.append(file)
tracked_and_not_wanted = sorted(tracked_and_not_wanted)
```
